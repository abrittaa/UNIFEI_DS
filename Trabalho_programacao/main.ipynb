```python
from CRUD import adicionar_produto, remover_produto, atualizar_produto, buscar_produto
from inventario import dictionary, listar_produtos
from parte4codigo import exibir_relatorio, salvar_inventario, carregar_inventario
```

```python
def exibir_menu():
    print("\n===== SISTEMA DE INVENTÁRIO =====")
    print("1. Adicionar produto")
    print("2. Remover produto")
    print("3. Atualizar produto")
    print("4. Buscar produto (nome ou ID)")
    print("5. Listar produtos")
    print("6. Exibir relatório")
    print("7. Salvar inventário (CSV)")
    print("8. Carregar inventário (CSV)")
    print("0. Sair")
    print("=================================")

def main():
    produtos = dictionary["inventario"]  # aponta para a lista do inventario.py

    while True:
        exibir_menu()
        opcao = input("Escolha uma opção: ").strip()

        if opcao == "1":
            adicionar_produto(produtos)
        elif opcao == "2":
            remover_produto(produtos)
        elif opcao == "3":
            atualizar_produto(produtos)
        elif opcao == "4":
            buscar_produto(produtos)
        elif opcao == "5":
            listar_produtos()
        elif opcao == "6":
            print(exibir_relatorio(produtos))
        elif opcao == "7":
            print(salvar_inventario(produtos))
        elif opcao == "8":
            carregados = carregar_inventario()
            produtos.clear()
            produtos.extend(carregados)
            print(f"{len(carregados)} produto(s) carregado(s).")
        elif opcao == "0":
            print("Encerrando o sistema. Até logo!")
            break
        else:
            print("Opção inválida! Tente novamente.")

if __name__ == "__main__":
    main()
```
